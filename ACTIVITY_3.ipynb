```python
import numpy as np

# Step 1: Model a three-node resistive circuit (nodal-voltage analysis).
# Rows/columns represent nodes 1, 2, 3. Each diagonal entry is the total
# conductance (S) connected to that node; off-diagonal entries are the
# negative conductance shared between two nodes. The right-hand side is
# the current (A) injected into each node.
conductance_matrix_s = np.array([
    [ 0.50, -0.20, -0.10],
    [-0.20,  0.45, -0.15],
    [-0.10, -0.15,  0.35],
])
injected_current_a = np.array([2.0, 1.2, 0.8])

print("Coefficient matrix shape:", conductance_matrix_s.shape)
print("Right-hand side shape:", injected_current_a.shape)

# Step 2: Report determinant, rank, and condition number before solving.
determinant = np.linalg.det(conductance_matrix_s)
rank = np.linalg.matrix_rank(conductance_matrix_s)
condition_number = np.linalg.cond(conductance_matrix_s)

diagnostics = {
    "determinant": float(round(determinant, 6)),
    "rank": int(rank),
    "condition_number": round(float(condition_number), 3),
}

print("Diagnostics:", diagnostics)

# Step 3: Solve directly with `solve`, never with an explicit inverse.
node_voltage_v = np.linalg.solve(conductance_matrix_s, injected_current_a)
print("Node voltages (V):", np.round(node_voltage_v, 4))

# Step 4: Verify by substitution — recompute A @ x and compare to b.
residual = conductance_matrix_s @ node_voltage_v - injected_current_a
max_abs_residual = float(np.abs(residual).max())

assert np.allclose(residual, 0)

report = {
    "node_voltages_v": np.round(node_voltage_v, 4).tolist(),
    "determinant": diagnostics["determinant"],
    "rank": diagnostics["rank"],
    "condition_number": diagnostics["condition_number"],
    "max_abs_residual_a": max_abs_residual,
}
report
```

```python
import numpy as np

# Declare the conductance as 3x3 matrix and an injected_current_array with 3 columns
conductance_matrix_s = np.array([
    [ 0.50, -0.20, -0.10],
    [-0.20,  0.45, -0.15],
    [-0.10, -0.15,  0.35],
])

injected_current_a = np.array([2.0, 1.2, 0.8])

# calculate determinant, rank, and condition
determinant = np.linalg.det(conductance_matrix_s)
rank = np.linalg.matrix_rank(conductance_matrix_s)
condition_num = np.linalg.cond(conductance_matrix_s)

# find the volate on each nodes
node_volatages = np.linalg.solve(conductance_matrix_s, injected_current_a)

# verify values using substituion
residual = conductance_matrix_s @ node_volatages - injected_current_a
abs_max_residual = float(np.abs(residual).max())

print("residual: ", residual)

assert np.allclose(residual, 0)

report = {
    "Diagnosis": {
        "Determinant" : float(determinant),
        "Rank" : int(rank),
        "Condition" : float(np.round(condition_num, 2)),
    },
    "Node voltages" : np.round(node_volatages, 2).tolist(),
    "Absolute Maximum Residual": abs_max_residual
}

report
```
